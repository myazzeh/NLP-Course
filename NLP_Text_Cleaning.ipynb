```python
texts = [
    "مرحباااا!!! 😀😀 هذا اختبار.....!!! http://example.com   تواصل: +962-79-123-4567",
    "<div>مرحبا بكم في <b>الموقع</b></div> زورونا على www.site.com #ترحيب",
    "أرسل ايميل إلى test_user@example.com أو إلى: info@uni.edu.jo!!!",
    "@ahmed مرحبا #خبر_عاجل !!!--- مين موجود؟ :)",
    "راااااااائع😍😍😍 والله والله!!!",
    "السَّلامُ عَلَيْكُمْ، كَيْفَ حَالُكَ؟!!!",
    "هذا مشروع AI في جامعة UoJ 2025 - version 2.0 (beta)",
    "SELECT * FROM users; -- هجوم ؟؟ <script>alert('x')</script>",
    "مساء الخير����!! ما هذا؟؟؟؟؟؟؟",
    "مرحبا\n\n\nهذا نص يحتوي     على مسافات زائدة \t و أسطر\nمتعددة."
]
```

```python
import re

def clean_arabic_text(text):
    # Remove URL
    text = re.sub(r'http\S+|www\.\S+', '', text)
    # Remove Emails
    text = re.sub(r'\S+@\S+\.\S+', '', text)
    # Remive mentions
    text = re.sub(r'@\w+', '', text)
    # Remove Hashtags
    text = re.sub(r'#\w+', '', text)
    # Remove HTML tags
    text = re.sub(r'<.*?>', '', text)
    # Remove Numbers
    text = re.sub(r'\d+', '', text)
    # Remove SQL
    text = re.sub(r'SELECT.*?;|<script.*?</script>', '', text, flags=re.IGNORECASE)
    # Char deduplications
    text = re.sub(r'(.)\1{2,}', r'\1', text)
    # Remove Emoji
    text = re.sub(r'[^\w\s\u0600-\u06FF]', '', text)
    # Remove spaces
    text = re.sub(r'\s+', ' ', text).strip()
    return text
```

```python
#Clean Text
cleaned_text=[]
for i, txt in enumerate(texts, 1):
    cleaned = clean_arabic_text(txt)
    cleaned_text.append(cleaned)
    print(f"Sample {i}:")
    print(f"Before Cleaning : {txt}")
    print(f"After Cleaning : {cleaned}\n{'-'*60}\n")
cleaned_text
```
